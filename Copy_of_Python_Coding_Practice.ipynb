7. Take a float input from the user, and print it rounded to 2 decimal places, cast as a string, along with its data type before and after casting.


```python
num=float(input("enter a decimal number: "))
print("before casting:")
print("value:",round(num,2))
print("data type:",type(num))
str_num = str(round(num,2))
print('after casting:')
print("value:",str_num)
print("data_types:",type(str_num))
```
